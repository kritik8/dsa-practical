```python
from pymongo import MongoClient
client = MongoClient("mongodb+srv://jkritik8_db_user:g5wf1AGIVlljhGsh@dsa-practical.goruzzs.mongodb.net/")
col = client["sample_mflix"]["movies"]

# Connect check
print(client.list_database_names())

# CREATE
col.insert_many([{"title": "Demo 1", "year": 2024, "genres": ["Drama"], "imdb": {"rating": 7.5}},
                 {"title": "Demo 2", "year": 2025, "genres": ["Comedy"], "imdb": {"rating": 6.8}}])

# READ: filter, projection, sort
for m in col.find({"year": {"$gte": 2000}, "imdb.rating": {"$gt": 8}},
                  {"title": 1, "year": 1, "imdb.rating": 1, "_id": 0}).sort("imdb.rating", -1).limit(5):
    print(m)

# UPDATE
col.update_one({"title": "Demo 1"}, {"$set": {"imdb.rating": 8.0}})

# DELETE
col.delete_one({"title": "Demo 2"})

# AGGREGATION: movies per genre
for r in col.aggregate([{"$unwind": "$genres"},
                        {"$group": {"_id": "$genres", "count": {"$sum": 1}}},
                        {"$sort": {"count": -1}}, {"$limit": 10}]):
    print(r)

# AGGREGATION: average rating per year
for r in col.aggregate([{"$match": {"imdb.rating": {"$type": "number"}}},
                        {"$group": {"_id": "$year", "avg_rating": {"$avg": "$imdb.rating"}}},
                        {"$sort": {"_id": -1}}, {"$limit": 10}]):
    print(r)
```
